# A-R2: small MLP action scorer

Team A | Rishitha Gogineni | 21 November 2026

This is the completed Colab CPU run. A-R2 uses the same 375 features, normalization and game groups as A-R1. The dataset contains 2,161 decisions from 210 training games, 539 decisions from 45 validation games and 390 decisions from 45 test games. No new collection or split was made.

The network has one hidden layer with 32 tanh units and 12,064 parameters. Validation loss selected L2 equal to 0.0001 and the checkpoint from epoch 60. The learned scorer handles supported MAIN decisions; other selections use the frozen heuristic.

Test agreement with a highest scoring teacher action was 99.49 percent for A-R2 and 98.72 percent for A-R1. Exact choice agreement was 95.13 percent and 96.15 percent respectively. Tied teacher scores allow different highest scoring choices, so these agreement measures have different meanings.

The manual Kaggle upload was reported after the run. Its official status and rating have not been supplied. The original JSON and CSV outputs retain the status saved before uploading.

```python
import ast
import csv
import hashlib
import json
import os
import platform
import shutil
import subprocess
import sys
import tarfile
import traceback
import uuid
import zipfile
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
import matplotlib
import matplotlib.pyplot as plt

assert platform.system() == 'Linux', 'Use a fresh Colab CPU session.'
HIDDEN_UNITS = 32
MAX_EPOCHS = 60
N_EVALUATION_GAMES = 40
N_ARCHIVE_GAMES_PER_OPPONENT = 2
PROCESS_TIMEOUT_SECONDS = 1800
assert min(HIDDEN_UNITS, MAX_EPOCHS, N_EVALUATION_GAMES, N_ARCHIVE_GAMES_PER_OPPONENT) > 0
RUN_ID = datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ') + '_' + uuid.uuid4().hex[:8]
work_dir = Path('/content/ptcg_ar2') / RUN_ID
output_dir = work_dir / 'outputs'
source_dir = output_dir / 'source'
reference_dir = output_dir / 'reference_ar1'
sdk_dir = work_dir / 'official_sample'
random_dir = work_dir / 'random_agent'
teacher_dir = work_dir / 'heuristic_agent'
ar1_dir = work_dir / 'ar1_agent'
candidate_dir = work_dir / 'ar2_agent'
for directory in [source_dir, reference_dir, sdk_dir, random_dir, teacher_dir, ar1_dir, candidate_dir]:
    directory.mkdir(parents=True, exist_ok=False)

def sha256(path):
    digest = hashlib.sha256()
    with path.open('rb') as handle:
        for block in iter(lambda: handle.read(1024*1024), b''):
            digest.update(block)
    return digest.hexdigest()

def save_json(path, value):
    path.parent.mkdir(parents=True, exist_ok=True)
    temporary = path.with_suffix(path.suffix + '.tmp')
    temporary.write_text(json.dumps(value, indent=2) + '\n', encoding='utf-8')
    temporary.replace(path)

run_info = {'run_id': RUN_ID, 'model_id': 'A-R2', 'parent_run_id': '20261122T002608Z_bd91a79d',
            'python': sys.version, 'platform': platform.platform(), 'numpy': np.__version__,
            'matplotlib': matplotlib.__version__, 'hidden_units': HIDDEN_UNITS, 'activation': 'tanh',
            'maximum_epochs': MAX_EPOCHS, 'training_seed': 42, 'learning_rate': 0.003,
            'evaluation_battles_per_comparison': N_EVALUATION_GAMES, 'unit': 'individual_battle',
            'native_engine_seed_verified': False, 'feature_schema': 'ar1-visible-v1'}
save_json(output_dir / 'run_info.json', run_info)
print('Run:', RUN_ID)
print('Runtime:', sys.version.split()[0], platform.machine(), '| CPU')
```

## 3. Reuse the game groups

Keep the original 210 training games, 45 validation games and 45 test games. The test data was already inspected for A-R1. Reusing it here supports the planned architecture comparison; it is not a new unseen benchmark.


```python
manifest = json.loads((reference_dir / 'training/dataset_manifest.json').read_text())
groups, row_keys = [], set()
for split in ['train', 'validation', 'held_out']:
    group = manifest[split]
    ids = set(group['game_ids'])
    rows = [json.loads(line) for line in (reference_dir / 'training' / (split + '.jsonl')).read_text().splitlines() if line.strip()]
    assert len(rows) == group['rows'] and len(ids) == group['games']
    assert {row['game_id'] for row in rows} == ids
    for row in rows:
        key = row['game_id'], row['step']
        assert key not in row_keys
        row_keys.add(key)
    groups.append(ids)
    print(split, '| games:', group['games'], '| decisions:', group['rows'])
assert not (groups[0]&groups[1] or groups[0]&groups[2] or groups[1]&groups[2])
save_json(output_dir / 'input_data_check.json', {'passed': True, 'groups_reused': True,
          'decisions': len(row_keys), 'split_manifest_sha256': sha256(reference_dir / 'training/dataset_manifest.json')})
```

```python
package_ready = False
(output_dir / 'submission.tar.gz').unlink(missing_ok=True)
package_manifest = {'model_id': 'A-R2', 'built': False, 'package_ready': False, 'kaggle_status': 'not_submitted'}
archive_results = {}
if evaluation_passed:
    try:
        candidate_archive = work_dir / 'ar2_candidate.tar.gz'
        expected_files = {'main.py', 'model.json', 'deck.csv'}
        with tarfile.open(candidate_archive, 'w:gz') as archive:
            for name in sorted(expected_files):
                archive.add(candidate_dir / name, arcname=name, recursive=False)
        unpacked = work_dir / 'unpacked_submission'
        unpacked.mkdir()
        with tarfile.open(candidate_archive, 'r:gz') as archive:
            members = archive.getmembers()
            assert len(members) == 3 and {member.name for member in members} == expected_files
            for member in members:
                assert member.isfile() and not member.issym() and not member.islnk()
                content = archive.extractfile(member).read()
                assert content == (candidate_dir / member.name).read_bytes()
                (unpacked / member.name).write_bytes(content)
        package_manifest.update(built=True, sha256=sha256(candidate_archive), bytes=candidate_archive.stat().st_size,
                                files={name: sha256(unpacked / name) for name in sorted(expected_files)})
        for index, (label, opponent) in enumerate([('random', random_dir), ('heuristic', teacher_dir), ('ar1', ar1_dir)]):
            passed, report, summary = run_battles('archive_vs_' + label, unpacked, opponent,
                                                N_ARCHIVE_GAMES_PER_OPPONENT, 70000 + 100*index)
            archive_results[label] = {'passed': passed, 'summary': summary}
        package_ready = all(item['passed'] for item in archive_results.values())
        if package_ready:
            target = output_dir / 'submission.tar.gz'
            shutil.copy2(candidate_archive, target)
            assert sha256(target) == package_manifest['sha256']
    except Exception as error:
        package_ready = False
        (output_dir / 'submission.tar.gz').unlink(missing_ok=True)
        package_manifest['error'] = {'type': type(error).__name__, 'message': str(error), 'traceback': traceback.format_exc()}
        print('Package check failed:', str(error))
else:
    print('Packaging skipped because a required check did not pass.')
package_manifest.update(package_ready=package_ready, archive_checks=archive_results)
save_json(output_dir / 'package_manifest.json', package_manifest)
print('Package ready:', package_ready)
if package_ready:
    print('Archive:', 'submission.tar.gz', '| bytes:', package_manifest['bytes'], '| SHA 256:', package_manifest['sha256'])
```
